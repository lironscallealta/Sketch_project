4. Calcule e interprete las medidas de tendencia central de las notas del examen.

```python
#esta seria pregunta de prueba
media_examen = df_1['Nota Examen'].mean() #es el promedio
mediana_examen = df_1['Nota Examen'].median() # el valor de al medio
moda_examen = df_1['Nota Examen'].mode() #la moda moda no se puede ocupar con 1f
print(f"La media del examen es: {media_examen:.1f}")
print(f"La mediana del examen es: {mediana_examen:.1f}")
print(f"La moda del examen es: {moda_examen}")
```

5. ¿Entre qué valores se encuentra el 60% de las notas más bajas en la prueba 1?

Esto es medida tendencia central
esta seria la media moda y mediana
y la no central es el percentil

```python
p_60 = df_1['Prueba 1'].quantile(0.6)
min_p1 = df_1['Prueba 1'].min()
max_p1 = df_1['Prueba 1'].max()
print(f"El percentil 60 de la prueba 1 es:{min_p1} y {p_60:.1f}")
```
